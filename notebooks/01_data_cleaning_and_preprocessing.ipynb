```python
# 1. Convert the text strings into native datetime objects
# (Pandas is usually smart enough to infer the YYYY-MM-DD format automatically)
df['date'] = pd.to_datetime(df['date'])

# 2. Extract the clean, 4-digit year from the date and overwrite the messy season column
df['season'] = df['date'].dt.year

# 3. Verify the changes
print("--- Cleaned Seasons ---")
print(df['season'].unique())
print(f"\nDate column type: {df['date'].dtype}")
```

```python
# Sort from 2008 to present, and from Ball 1 to the final wicket
df = df.sort_values(by=[
    'date',      # 1st: The calendar day
    'match_id',  # 2nd: To separate afternoon vs evening games on the same day
    'innings',   # 3rd: 1st innings before 2nd innings
    'over',      # 4th: Over 0 to Over 19
    'ball'       # 5th: Ball 1 to Ball 6 (or 7/8 for extras)
])

# Reset the index so row 0 is truly the very first ball in IPL history
df = df.reset_index(drop=True)

# Verify the first ball ever bowled
df.head(1)
```

```python
print(df['innings'].unique())
```

```python
# Fill empty wicket fields with logical text
df['wicket_kind'] = df['wicket_kind'].fillna('No Wicket')
df['wicket_player_out'] = df['wicket_player_out'].fillna('Nobody')
```

```python
# Step 1: Fill nulls with 0
df['win_by_runs'] = df['win_by_runs'].fillna(0)
df['win_by_wickets'] = df['win_by_wickets'].fillna(0)

# Step 2: Cast to integers (since you can't win by 2.5 runs)
df['win_by_runs'] = df['win_by_runs'].astype(int)
df['win_by_wickets'] = df['win_by_wickets'].astype(int)
```

```python
# Tag unawarded matches clearly
df['player_of_match'] = df['player_of_match'].fillna('No Result')
```

```python
# Combine all unique names from batters and bowlers into one master list
all_players = pd.concat([df['batter'], df['bowler']]).unique()

# Sort them alphabetically
all_players_sorted = np.sort(all_players)

# Print the first 50 to see what we are dealing with
print(all_players_sorted[:50])
```

```python
from thefuzz import process

# 1. Gather every unique player name from all relevant columns
player_cols = ['batter', 'bowler', 'non_striker', 'player_of_match', 'wicket_player_out']
all_names = pd.concat([df[col] for col in player_cols]).dropna().unique()

# 2. Create a list to store our potential typos
potential_typos = []
processed_names = set() # To avoid checking the same name twice

print("Scanning for similar names...")

for name in all_names:
    if name not in processed_names:
        # Extract the top 3 matches for 'name' from the master list
        similar_names = process.extract(name, all_names, limit=4)
        
        # Keep matches that score between 85 and 99 
        # (100 is an exact match, below 85 is usually a different person)
        high_matches = [match for match in similar_names if 87 <= match[1] < 100]
        
        if high_matches:
            print(f"\nTarget Name: {name}")
            for match_name, score in high_matches:
                print(f" -> Found: '{match_name}' (Score: {score})")
                processed_names.add(match_name) # Mark as processed
                
        processed_names.add(name)
```

```python
# The meticulously vetted correction dictionary
fuzzy_corrections = {
    'Arshad Khan (2)': 'Arshad Khan',
    'NA Saini': 'N Saini',
    'R Bishnoi': 'Ravi Bishnoi',
    'M.S. Dhoni': 'MS Dhoni',
    'M S Dhoni': 'MS Dhoni'
}

# Apply the corrections across all player columns
player_cols = ['batter', 'bowler', 'non_striker', 'player_of_match', 'wicket_player_out']

for col in player_cols:
    df[col] = df[col].replace(fuzzy_corrections)

print("Player names successfully vetted and sanitized with 100% accuracy!")
```

```python
# Step 1: Ensure the directory exists (just in case you haven't created the folder yet)
output_dir = Path('../data/processed')

# Step 2: Define the exact file path
output_file = output_dir / 'ipl_cleaned.csv'

# Step 3: Export the DataFrame
# VERY IMPORTANT: index=False prevents pandas from writing the row numbers as a new column
df.to_csv(output_file, index=False)

print(f"Success! Clean data saved to: {output_file}")
```
