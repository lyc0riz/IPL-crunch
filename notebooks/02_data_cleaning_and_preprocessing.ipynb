```python
# Combine team1 and team2 columns, get unique values, and drop any nulls
unique_teams = pd.concat([df['team1'], df['team2']]).dropna().unique()

# Sort them alphabetically to  spot variations 
unique_teams.sort()

# Print the list
print(f"Total Unique Team Entries Found: {len(unique_teams)}\n")
print("--- Team List ---")
for team in unique_teams:
    print(team)
```

```python
df = proc.standardize_teams(df)
# changes team names to current ones
# 'Delhi Daredevils': 'Delhi Capitals'
# Kings XI Punjab': 'Punjab Kings'
# 'Royal Challengers Bangalore': 'Royal Challengers Bengaluru'
#'Rising Pune Supergiants': 'Rising Pune Supergiant'
```

```python
# Get unique cities, drop nulls, and sort alphabetically
unique_cities = df['city'].dropna().unique()
unique_cities.sort()

print(f"Total Unique Cities Found: {len(unique_cities)}\n")
print("--- City List ---")
for city in unique_cities:
    print(city)
```

```python
df = proc.standardize_cities(df)
# fix city names
# 'Bangalore': 'Bengaluru',
# 'Mohali': 'Chandigarh',
# 'New Chandigarh': 'Chandigarh'
```

```python
# Get unique venues, drop nulls, and sort alphabetically
unique_venues = df['venue'].dropna().unique()
unique_venues.sort()

print(f"Total Unique Venues Found: {len(unique_venues)}\n")
print("--- Venue List ---")
for venue in unique_venues:
    print(venue)
```
